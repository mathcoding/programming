# Strutture dati gerarchiche
Abbiamo visto come una coppia permette di ottenere un unico dato aggregato partendo da due dati primitivi.



````python
a = (3, 4)
print(a)
````

Concettualmente, una coppia viene rappresentata in modo diretto usando una raffigurazione *box-and-pointer* (VEDI SLIDES). Per una coppia in pratica abbiamo un doppio box di "puntatori", ognuno che "punta" ad uno dei due dati primitivi.

La coppia pu&ograve; essere definita anche in termini di se stessa: possiamo avere una coppia di coppie:

````python
a = ((1,2), (3,4))
````

````python
print(a)
````

La possibilit&agrave; di definire coppie i cui elementi sono delle coppie &egrave; essenziale per poter rappresentare una **LISTA** di dati, in cui ciascun elemento della lista pu&ograve; essere lui stesso una lista.

Una delle strutture dati pi&ugrave; utilizzate che possiamo costruire a partire dalle coppie &egrave; quella di **SEQUENZA**, ovvero di una collezione ordinata di dati. Ovviamente, possiamo avere molti modi di rappresentare una sequenza di numeri in termini di coppie, e il modo pi&ugrave; semplice &egrave; quello di usare una *catena di coppie*, in cui l'ultima coppia ha come secondo elemento un **simbolo** che rappresenta un valore il cui significato &egrave; di segnalare la fine della lista.

````python
Ls = (1, (2, (3, (4, None))))
print(Ls)
````

La sequenza di numeri ordinati in `Ls` &egrave; comunemente chiamata una *lista*. Il termine *lista* in base al linguaggio di programmazione usato potrebbe introdurre alcune ambiguit&agrave; che vedremo pi&ugrave; avanti.

## Definizione di PairList
Pi&ugrave; formalmente, possiamo definire una lista nel modo seguente.

Una **LISTA** &egrave;:

1. Una **lista vuota**, rappresentata concretamente dal valore `None`

2. Una coppia di valori, il cui **primo elemento** rappresenta un elemento della lista, e il **secondo elemento** rappresenta la lista dei valori che seguono il primo elemento


**ESEMPIO:** Procedura che restituisce una lista di numeri naturali da 1 a $n$.

````python
def MakeRange(n):
    def MakeI(a):
        if a > n:
            return None
        return (a, MakeI(a+1))
    return MakeI(1)
````

````python
Ls = MakeRange(5)
print(Ls)
````

Il valore `None` in questo contesto pu&ograve; essere pensato come ad una sequenza di nessun elemento, ovvero una lista vuota.

Essendo una lista una catena di coppie, possiamo sempre usare i *selettori* di una coppia per accedere a quella che viene chiamata la testa (head) e la coda (tail) di una lista:

````python
def Head(Ls):
    return Ls[0]

def Tail(Ls):
    return Ls[1]

Ls = MakeRange(5)
print("head:", Head(Ls))
print("tail:", Tail(Ls))
````

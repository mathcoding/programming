## 4.1 La rappresentazione di Numeri Razionali
Vogliamo ora definire un sistema di calcolo per i numeri razionali.

Per iniziare, possiamo definire un'operazione `AddQ(x,y)` che prende in input due numeri razionali, per esempio $\frac{1}{2}$ e $\frac{2}{3}$, e restituisce la loro somma (nell'esempio, $\frac{1}{2}+\frac{2}{3}=\frac{7}{6})$. In termini di dati primitivi, un numero razionale può essere visto come una **COPPIA** di due numeri interi: un **numeratore** e un **denominatore**.

In pratica, possiamo definire un sistema di calcolo in cui ogni numero razionale viene definito da due numeri interi. Per poter calcolare la somma di due numeri razionali dobbiamo scrivere almeno due procedure, la prima per calcolare il numeratore e la seconda per calcolare il denominatore. Tuttavia, procedendo con questo approccio si arriverebbe presto ad un codice complicato da

1. Leggere 
2. Mantenere
3. Migliorare

Questo perché dovremmo sempre tenere traccia di quale numeratore corrisponde a quale denominatore.

Sarebbe più semplice poter in qualche modo "incollare" un numeratore al suo denominatore, per ottenere un nuovo tipo di dato, in modo che le nostre funzioni possano elaborare direttamente il nuovo **TIPO** di dato aggregato.

In questo modo, sarebbe più semplice aumentare la modularità dei nostri programmi. Se possiamo manipolare i numeri razionali direttamente come degli **oggetti** a sé stanti, come dei dati primitivi, allora potremmo separare nei nostri programmi il codice che usa i numeri razionali, dalla parte di codice che specifica come eseguire le operazioni aritmetiche sui numeri razionali stessi.

L'uso di *dati aggregati* (o *dati composti*) porta ad un linguaggio ancora più **espressivo** di quello studiato sino ad ora, attraverso un meccanismo che viene chiamato di *data abstraction*.

**ESEMPIO:** Si consideri l'idea di formare una combinazione lineare $ax + by$. Possiamo scrivere una procedura che prende in input quattro parametri `a, b, x, y` e restituisce il valore `ax + by`. Se gli argomenti dati in input sono dei numeri, allora possiamo scrivere:

```python
def CombinazioneLineare(a, b, x, y):
    return a*x + b*y
```

Supponiamo per&ograve; che non vogliamo definire una combinazione lineare solo di numeri interi o di numeri in virgola mobile, ma vogliamo esprimere una combinazione lineare per qualsiasi tipo di dato per cui sono state definite le operazioni `Add` e `Mul`, come ad esempio, i numeri razionali, i numeri complessi, o i polinomi. In generale, potremmo scrivere:

```python
def CombinazioneLineare(a, b, x, y):
    return Add(Mul(a, x), Mul(b, y))
```

In cui `Add` e `Mul` non sono necessariamente le operazioni aritmetiche di addizione `+` e moltiplicazione `*`, ma possono essere funzioni pi&ugrave; complesse che eseguono l'operazione corretta per qualsiasi tipo di dato che gli viene passato in input. Il punto cruciale &egrave; che l'unica cosa che la procedura `CombinazioneLineare(a, b, x, y)` deve sapere dei parametri formali `a, b, x, y`, &egrave; che le due procedure `Add` e `Mul` eseguiranno su quei dati le operazioni in modo corretto. Dalla prospettiva della procedura `CombinazioneLineare` &egrave; irrilevante cosa siano effettivamente `a, b, x, y` e di come siano effettivamente rappresentati i loro valori in termini di dati primitivi.

Per rendere l'idea concreta, vediamo ora come implementare un sistema di funzioni per eseguire le operazioni aritmetiche sui numeri razionali.

## 4.2 Dati astratti e dati concreti
L'idea base del *data abstraction* &egrave; di strutturare i programmi che devono usare dei dati aggregati in modo tale che questi possano operare direttamente su **dati astratti**. In pratica, non vogliamo fare nessuna ipotesi su come siano effettivamente memorizzati i dati aggregati, ma ci interessano di più le loro proprietà. Allo stesso tempo un **dato concreto** &egrave; implementato in modo indipendente dai programmi che useranno quel tipo di dato aggregato. L'interfaccia tra i dati astratti e i dati concreti sar&agrave; costituita da un insieme di funzioni, chiamate **selettori** e **costruttori**, che implementano i *dati astratti* in termini della loro vera rappresentazione *concreta* in termini di dati primitivi o di altri dati aggregati.

Supponiamo di avere gi&agrave; una funzione per definire un numero razionale a partire da un numeratore e un denominatore. Questa funzione viene chiamata il **costruttore** del tipo di dato. Si noti come una coppia sia un oggetto, un unico dato, a cui possiamo assegnare un nome tramite un'operazione di assegnamento, e che possiamo utilizzare *come se fosse un dato primitivo*.

A questo punto, possiamo usare una *coppia* come elemento costitutivo delle funzioni definite sopra per i numeri razionali:

```python
# Costruttore
def MakeQ(n, d=1):
    return (n,d)

# Selettore per il numeratore
def Num(x):
    return x[0]

# Selettore per il denominatore
def Den(x):
    return x[1]
```

### 4.2.2 Definizione funzionale di coppia
Consideriamo la **coppia (pair)**, che sino ad ora abbiamo considerato come un dato primitivo del linguaggio.

Se la coppia non fosse direttamente definita in un linguaggio di programmazione, ma se avessimo il modo di definire funzioni, allora potremmo dare una definizione interamente funzionale di coppia, utilizzando la programmazione funzionale **High Order**. Una coppia può essere vista come una funzione che prende in input un valore pari a 1 oppure a 2 (l'indice della coppia), e restituisce il primo o il secondo elemento. Questo può essere implementato in Python con la sintassi seguente: 

```python
def MakePair(x, y):
    def Select(i):
        if i == 1:
            return x
        if i == 2:
            return y
        print("Errore: in una coppia sono presenti solo due elementi.")
    
    return Select
```

```python
a = MakePair(2,3)
```

A questo punto possiamo cambiare la nostra implementazione di numero razionale, senza dover cambiare le operazioni aritmetiche definite sui numeri razionali:

```python
def MakeQ(n, d):
    return MakePair(n, d)

def Num(x):
    return x(1)

def Den(x):
    return x(2)
```

### 4.2.3 Approfondimento: *Lambda Calculus*
Da un punto di vista teorico, in un linguaggio di programmazione che pu&ograve; manipolare le funzioni come se fossero dei dati, non abbiamo pi&ugrave; bisogno di dati primitivi per rappresentare i numeri (almeno se ci limitiamo ai numeri interi non negativi): tutto quello che ci serve &egrave; di definire lo zero e l'operazione per aggiungere 1 ad un numero:

```python
Zero = lambda f: lambda x: x

def AddOne(n):
    return lambda f: lambda x: f(n(f)(x))
```

Questa rappresentazione &egrave; dovuta ad [Alonzo Church](https://en.wikipedia.org/wiki/Alonzo_Church), il logico matematico che ha inventato il [*Lambda Calculus*](https://en.wikipedia.org/wiki/Lambda_calculus), che sta alla base di tutti i linguaggi di programmazione funzionale, come ad esempio [Haskell](https://www.haskell.org).

Per "vedere" il numero intero rappresentato da un numerale di Church, basta applicare $n$ volte la funzione $k \mapsto k+1$ partendo da $0$:

```python
def ToInt(n):
    return n(lambda k: k + 1)(0)

Uno = AddOne(Zero)
Due = AddOne(Uno)
print(ToInt(Zero), ToInt(Uno), ToInt(Due))
```

### 4.3.1 Operazioni sui numeri complessi
Sui numeri complessi si possono definire le operazioni seguenti:


**Modulo:**
$$z=a + bi, \quad |z| = \sqrt{a^2 + b^2}$$

**Argomento** (ricordando che $z=a + bi=|z|(\cos(\theta) + i \sin(\theta))$):

$$z=a + bi, \quad \theta=\arctan\left(\frac{b}{a}\right)\quad (\text{se } a > 0)$$

In generale, per tenere conto del quadrante in cui si trova $z$, conviene usare la funzione `atan2(b, a)` del modulo `math`.

**Coniugato**:
$$z=a + bi, \quad \bar{z} = a - bi$$

**Reciproco**:
$$z=a + bi, \quad z^{-1}=\frac{a - bi}{a^2 + b^2}$$

**Somma algebrica**:
$$z_1 + z_2 = (a,b) + (c,d) = (a+c, b+d)$$

**Prodotto**:
$$z_1*z_2 = (a,b)(c,d) = (ac - bd, bc+ad)$$

**Rapporto**:
$$\frac{z_1}{z_2} = \frac{(a,b)}{(c,d)} = \frac{(a,b)(c, -d)}{(c,d)(c, -d)} = \frac{(ac + bd, cb-ad)}{c^2 + d^2}$$

**Potenza**:
$$z^n = |z|^n(\cos(n\theta) + i \sin(n \theta))$$

**Esponenziale**:
$$e^z = e^a (\cos(b) + i \sin(b))$$

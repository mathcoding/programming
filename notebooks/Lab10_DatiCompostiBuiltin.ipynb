Il controllo dei tipi dei dati viene chiamato in gergo tecnico il **Type Checking**, e dipende dal [Type System](https://en.wikipedia.org/wiki/Type_system) definito dal linguaggio di programmazione che si sta usando. Il **Type System** di Python viene chiamato *dinamico* in quanto controlla i tipi dei dati delle variabili durante l'esecuzione del programma. Diversamente da altri linguaggi di programmazione, non è necessario specificare il tipo di dati direttamente nel codice.

In Python è possibile specificare nella definizione di una funzione il **tipo** dei parametri formali di input e il **tipo** della variabile di ritorno usando i [Type Hints](https://docs.python.org/3/library/typing.html). Attenzione, come la parola suggerisce (*hints*) non si tratta di richieste stringenti, ma di definizioni opzionali che possono facilitare la lettura del codice da altre persone, o possono essere utilizzati da alcuni programmi specifici, diversi dall'interprete di Python, per eseguire delle analisi statiche del codice.

**ESEMPIO**: Per esempio, una funzione che calcola l'ipotenusa di triangoli con lati di lunghezza rappresentata da numeri interi, potrebbe essere scritta nel modo seguente:

````python
# Importa la funzione per il calcolo della radice quadrata
from math import sqrt

# Usa i "type hints" per specificare i tipi in input e output
def Pitagora(a: int, b: int) -> float:
    return sqrt(a**2 + b**2)

print(Pitagora(3.4, 5))  # Funziona anche se 3.4 non è un int: i type hints non sono vincolanti
````

In questo corso, non faremo uso dei *type hints*.

I **SELETTORI** delle stringhe sono i seguenti.

* Si può accedere ai singoli caratteri di una stringa attraverso i **SELETTORI di INDICIZZAZIONE**, che sono definiti con le parentesi quadre `[i]`. Si ricordi che una stringa è una sequenza ordinata di caratteri: per esempio, data una stringa di *n* caratteri, il primo carattere è associato all'indice 0, e l'ultimo carattere è l'*(n-1)*-esimo carattere della stringa. Ad esempio:

````python
tmp = "banda"
print(tmp[0], tmp[1])
````

````python
tmp[7]
````

````python
print(tmp[-1], tmp[-2])
````

* L'ultimo esempio mostra implicitamente che le stringhe sono delle strutture dati sequenziali indirizzate in modo circolare: se si usa un numero negativo nell'operatore di indicizzazione, si accede all'*i*-esimo elemento a partire dalla fine della stringa. Tuttavia, se cerchiamo di indicizzare un elemento che eccede la lunghezza della stringa, otteniamo un **IndexError: string index out of range**:

````python
tmp[6]
````

* È possibile estrarre delle sotto sequenze con un'operazione di **SLICING**. Se `s` è una stringa, l'espressione `s[start:end]` restituisce la sotto stringa di `s` che inizia all'indice `start` e termina all'indice `end-1`. Per convenzione, il secondo indice `end` è escluso dall'operazione di slicing, in modo tale che l'espressione `s[0:len(s)]` abbia il valore che uno si aspetta. Quest'ultima espressione è equivalente a `s[:]`. 
Se il primo indice viene omesso, di default assume il valore 0. Se il secondo indice viene omesso, di default assume il valore pari alla lunghezza della stringa.

````python
a = "Ma tutta la banda che suona e che canta"
print(a[3:8])
````

````python
# Primo indice omesso, di default 'start=0'
a[:8]
````

````python
# Secondo indice omesso, di default 'end=len(a)'
a[28:]
````

````python
a[1:2000] # Se 'end' supera la lunghezza della stringa, lo slicing si ferma all'ultimo carattere
````

Per le stringhe, i **simboli usati per le operazioni aritmetiche in notazione infissa (infix)** sono stati ridefiniti in modo tale che abbiano il significato seguente:

1. L'operatore di somma `+(a, b)` effettua la **CONCATENAZIONE** di due stringhe, equivalente alla funzione `Append(As, Bs)` che abbiamo definito per le PairsList.
2. L'operatore di moltiplicazione `*(a, b)`, se richiamato con un intero $n$ e una stringa, viene chiamato l'operatore di **RIPETIZIONE** e *ripete* (o concatena) la stringa $n$ volte.

In entrambi questi due casi, in termini tecnici, si dice che i due operatori di addizione e moltiplicazione sono stati **"overloaded"**: è stata definita una funzionalità diversa che dipende dai tipi di dati che vengono passati in input. Per l'addizione, è come se fosse stata definita la funzione seguente:

```
def add(a, b):
    if IsPairsList(a) and IsPairsList(b):
        return Append(a, b)
    if type(a) == int and type(b) == int:
        return a+b
    print('TypeError')
        
```
    
**NOTA**: Se si utilizza l'operatore di prodotto tra due stringhe, o si utilizza un qualsiasi operatore aritmetico che non sia stato ridefinito per le stringhe, si ottiene un `TypeError`:

````python
a*b
````

### 4.2.3 Le stringhe come dati di tipo **SEQUENZE**

Le stringhe in Python sono uno dei tipi di dati di Python che appartengono alla classe di dati **SEQUENZE**. In Python, esistono una serie di operazioni che sono definite nello stesso modo per tutti i tipi di dati che rappresentano delle sequenze. Queste operazioni sono:

* Gli operatori di **slicing** `[start:end]` che sono usati come selettori.
* La funzione `len(X)` che prende in input una sequenza e restituisce la lunghezza della sequenza, ovvero il numero di elementi contenuti nella sequenza. Ad esempio: `len('abc')` restituisce il valore 3. La funzione `len(X)` è analoga alla funzione `Length(Xs)` definita per le PairsList.
* Si può controllare se un dato elemento appartiene ad una sequenza usando la sintassi: `<elemento> in <sequenza>` che è un predicato che restituisce `True`, se l'elemento appare nella sequenza almeno una volta, e restituisce `False` altrimenti. Si noti che questa era esattamente la semantica del predicato `Contains(Ls, value)` definito per le PairsList.

````python
a = 'la vecchia che balla con il vecchio'
````

````python
'vecchio' in a
````

````python
# Distingue le maiuscole (case sensitive)
'V' in a
````

````python
'' in a
````

* Per semplificare la scrittura di procedure, per i tipi di dati che sono delle *sequenze*, è stata introdotta la **SINTASSI** seguente, che permette di accedere a ciascun elemento della sequenza, una alla volta:

```
for <elemento> in <sequenza>:
    <body>  # in cui si usa l'elemento
```

e quindi possiamo scrivere, per esempio:

````python
for x in 'abcde':
    print(x)
````

````python
1/3
````

1. `end`: specifica il carattere da usare per terminare la stampa della stringa. Di default è uguale al carattere di ritorno a capo `\n`.
2. `sep`: specifica il carattere da usare per separare più stringhe. Di default è uguale ad uno spazio.

````python
print('uno','due','tre', sep='\t', end='\n')
````

### 4.2.4 Funzioni built-in per le stringhe
Le funzioni seguenti implementano le operazioni fondamentali per le stringhe. Si noti che nessuna di queste funzioni modifica la stringa `s`: le funzioni che restituiscono una stringa restituiscono sempre una nuova stringa.

* `s.count(s1)`: conta quante volte la stringa `s1` è contenuta in `s`
* `s.find(s1)`: restituisce l'indice della stringa `s` in cui viene trovata per la prima volta la stringa `s1`; se la stringa `s1` non viene trovata viene restituito il valore -1
* `s.rfind(s1)`: come sopra, ma la ricerca inizia dalla fine della stringa `s` (la `r` sta per `reversed`)
* `s.lower()`: converte tutte le lettere in minuscolo
* `s.upper()`: converte tutte le lettere in maiuscolo
* `s.replace(old,new)`: sostituisce tutte le sotto stringhe uguali a `old` in `s` con la stringa `new`
* `s.strip()`: rimuove tutti i caratteri blanks iniziali e finali dalla stringa `s`
* `s.rstrip()`: rimuove tutti i caratteri blanks finali dalla stringa `s`
* `s.lstrip()`: rimuove tutti i caratteri blanks iniziali dalla stringa `s`
* `s.strip(c)`: rimuove i caratteri contenuti nella stringa `c` all'inizio e alla fine della stringa `s`
* `s.split(d)`: suddivide la stringa in sotto stringhe usando il carattere `d` come separatore

### 4.2.5 Conversione di caratteri in numeri interi (e viceversa)
In Python non esiste un tipo di dato specifico per identificare un singolo carattere: il singolo carattere è semplicemente una stringa di lunghezza pari a uno. Tuttavia è possibile convertire un carattere nel suo corrispondente [codice ASCII](https://en.wikipedia.org/wiki/ASCII) usando la funzione `ord(s)` in cui `s` è una stringa di lunghezza 1. Il codice ASCII rappresenta l'insieme dei 128 caratteri, con codici da 0 a 127, che possono essere rappresentati usando stringhe di 7 bits, in quanto $2^7=128$. 

![Codice ASCII](./images/asciiTable.png)

Usando 8 bits al posto di 7, si possono rappresentare altri 128 caratteri, che sono mostrati in [questa immagine](http://www.table-ascii.com/table_ascii_1.jpg).

## 4.6 Numeri complessi
Tra i principali dati **built-in** di Python esiste anche il tipo di dato *numero complesso*, composto da una coppia di numeri in virgola mobile che rappresentano la parte reale e la parte immaginaria.

Per poter costruire un dato di tipo **numero complesso** possiamo usare i seguenti **COSTRUTTORI**:

```python
# possiamo usare direttamente il carattere 'j' dopo un numero
c1 = 1 - 2j
# oppure utilizzare la funzione built-in "complex":
c2 = complex(1,2)
```

```python
print(c1, type(c1))
```

```python
print(c2, type(c2))
```

Per accedere ai singoli elementi del numero complesso, possiamo usare i seguenti **SELETTORI**:

```python
print('numero: {}, parte reale: {}, parte immaginaria: {}'
        .format(c1, c1.real, c1.imag))
```

```python
print('numero: {}, parte reale: {}, parte immaginaria: {}'
        .format(c2, c2.real, c2.imag))
```

Inoltre, è possibile ottenere il coniugato di un numero complesso attraverso la funzione `conjugate()`:

```python
print('numero: {}, coniugato: {}, parte reale: {}, parte immaginaria: {}'
        .format(c2, c2.conjugate(), c2.conjugate().real, c2.conjugate().imag))
```

```python
j=1
```

```python
1j
```

Si noti che `conjugate()` è una funzione che prende in modo implicito come argomento il numero stesso, con la notazione **dot.function()** che abbiamo già visto per le stringhe, le liste e i dizionari. Invece, `real` e `imag` sono degli **attributi** (tipi particolari di **selettori**) del dato composto `complex`.

Per i numeri complessi sono stati ridefiniti tutti gli operatori aritmetici, il predicato di uguaglianza, e il calcolo del modulo.

```python
a = 1+3j
b = -3 -5j
c = 1j
d = 2
print(a+b*c/d)
print(1==b)
print(abs(b))
```

```python
# Il modulo di un numero complesso a+bi è la radice quadrata di a^2 + b^2
from math import sqrt
print(b, sqrt((-3)**2 + (-5)**2))
```

Per passare alle coordinate polari di un numero complesso, si deve tuttavia importare la libreria `cmath`. In questa libreria, si hanno a disposizione le funzioni seguenti:

1. `phase(x)`: per ottenere la *fase* del numero complesso `x`, che è equivalente a `math.atan2(x.imag, x.real)`
2. `polar(x)`: per ottenere la rappresentazione di `x` in coordinate polari
3. `rect(r, phi)`: per ottenere il numero complesso `x` dalle coordinate polari `r` e `phi`, ovvero per ottenere il numero `r * (math.cos(phi) + math.sin(phi)*1j)`.

```python
from cmath import phase, polar, rect
phi = phase(c1)
print(phi)
```

```python
r, phi = polar(c1) # r, phi = abs(c1), phase(c1)
print(c1, r, phi)
```

```python
print(rect(r,phi))
```

La documentazione di Python contiene la [lista completa di funzioni per i numeri complessi](https://docs.python.org/3/library/cmath.html).

**NOTA:** La funzione `polar` restituisce una *tupla* di due valori, che nella cella precedente abbiamo assegnato a due variabili con un'unica istruzione (*unfolding*). Lo stesso meccanismo si può usare con le funzioni che scriviamo noi:

```python
def Terna(a):
    return (a, a**2, a**3)
b = Terna(3)
print(b, type(b))
```

## 4.7 Liste, Vettori e Matrici
Una lista di numeri può essere utilizzata per rappresentare un vettore di $\mathbb{R}^n$. In modo analogo, una matrice rettangolare di $n$ righe e $m$ colonne può essere rappresentata in Python utilizzando delle liste di liste che contengono gli elementi della matrice. Per esempio, la matrice

$A = \left[ \begin{array}{ccc} 1 & 3 & 0 \\ -1 & 1 & 9 \\ -3 & 4 & 1 \end{array} \right]$

può essere rappresentata dalla lista di liste (una lista per ogni riga, rappresentazione *[row-wise](https://en.wikipedia.org/wiki/Row-_and_column-major_order)*):

```python
A = [[1,3,0], [-1,1,9], [-3,4,1]]
```

```python
print(A)
```

oppure dalla lista di liste (una lista per ogni colonna, rappresentazione *[column-wise](https://en.wikipedia.org/wiki/Row-_and_column-major_order)*):

```python
B = [[1,-1,-3], [3,1,4], [0,9,1]]
print(B)
```

```python
# Accesso alla seconda riga
print(A[1])
# Oppure accesso alla seconda colonna
print(B[1])
```

```python
# Accesso al terzo elemento della seconda riga, ovvero elemento a(2,3)
print(A[1][2])
print(B[2][1])
```

La rappresentazione per righe di una matrice è ad oggi quella più usata nei linguaggi di programmazione che derivano dal C/C++. La rappresentazione per colonne è una delle caratteristiche principale del linguaggio per calcolo scientifico chiamato [Fortran](https://en.wikipedia.org/wiki/Fortran).

### 4.7.3 La *matrice* scacchiera 
L'esempio più semplice di immagine che corrisponde ad una matrice è sicuramente una scacchiera, in cui possiamo assegnare un numero al bianco e un altro numero al nero.

Per esempio, il codice seguente crea un'immagine corrispondente ad una scacchiera 8x8, in cui ad ogni coppia di numeri interi $(i,j)$ viene assegnato il valore 0, che viene *mappato* al bianco, oppure il valore 1, che viene *mappato* al nero. La mappatura al bianco e al nero viene determinata dal parametro `cmap=binary`.

```python
matrix([[(i+(j%2))%2 for i in range(8)] for j in range(8)])
```

```python
import matplotlib.pyplot as plt
from numpy import matrix

# NOTA: in alcuni casi in jupyter si deve aggiungere il comando seguente:
%matplotlib inline

def DrawChessBoard(n=8):
    # L'operatore '%' calcola il resto della divisione
    ChessBoard = matrix([[(i+(j%2))%2 for i in range(n)] for j in range(n)])
    #print(ChessBoard)
    # Crea una figura e disegna l'immagine data dalla matrice
    fig = plt.figure(figsize=(4,4))
    # Uso la colormap 'binary' per avere la scacchiera in bianco e nero
    img = plt.imshow(ChessBoard, cmap='binary')
    plt.show()
    
DrawChessBoard(8)
```

### 4.7.4 La *matrice* circolare dei numeri complessi
Per farsi un'idea di come vengono gestiti i colori, si consideri il codice seguente, in cui si usa il piano complesso, e si fa corrispondere ad ogni elemento $a_{rc}$ di una matrice quadrata $2n \times 2n$, con $r,c = 0,\dots,2n-1$, il valore:

$$a_{rc} = \mid complex(s\,(c-n),\, s\,(r-n)) \mid$$

in cui $s$ è il fattore di scala (il parametro `scale`), e `complex(a, b)` è la primitiva di Python che crea il numero complesso $a+b\,i$ (dove la lettera *i* rappresenta l'unità immaginaria), come abbiamo visto nella Sezione 4.6.

```python
# Calcola il modulo del numero complesso
# che viene usato per stabilire il colore nell'immagine
def ComplexModulo(z):
    return abs(z)

# Costruisci una matrice 2*n x 2*n dove ciascun elemento
# corrisponde ad un pixel dell'immagine; il valore numerico
# dell'elemento ne determina il colore
def MakeImage(F, n, scale=0.01):
    data = [i*scale for i in range(-n,n)]
    return matrix([[F(complex(a, b)) for a in data] for b in data])

# Disegna l'immagine della matrice creata dalla funzione F(z) : Z -> R
def DrawImage(F, n, scale):
    # Restituisce una matrice 2n x 2n
    m = MakeImage(F, n, scale)
    print(m)
    # Crea una figura, disegna l'immagine data dalla matrice, aggiungi la colorbar sulla destra
    plt.figure(figsize=(6,6))
    img = plt.imshow(m, extent=(-scale*n, scale*n, -scale*n, scale*n), 
                     cmap='jet') # jet, gray,
    plt.colorbar()
    plt.show()

# Alcuni parametri
DrawImage(ComplexModulo, 1000, 0.01)
```

## 4.3 Le tuple
Le tuple, come le stringhe, sono delle sequenze *non modificabili* di elementi, ovvero è un tipo di dato **read only**. A differenza delle stringhe non abbiamo nessun vincolo particolare sul **tipo** degli elementi che appaiono nella sequenza, che possono anche avere dei tipi tutti diversi tra loro.

Una possibile rappresentazione "box-and-pointer" di una tupla è la seguente:

<img src="images/tuplaBoxPointer.PNG" alt="Stringa BoxAndPointer" style="width: 400px;"/>

La "coppia" (pair) che abbiamo usato per costruire la nostra libreria `PairsList` è un caso particolare di tupla con lunghezza pari a due.

I metodi **COSTRUTTORI** per le tuple sono:

```
t1 = ()
t2 = tuple()
t3 = (1, 2.3, 'cool!', False)
```

I **SELETTORI** per il tipo di dato "tupla" sono gli operatori di indicizzazione e di slicing, che sono definiti con la stessa sintassi usata per le stringhe. 

````python
t = (1, 2.3, 'cool!', False)
print(t, type(t))
````

````python
print(t[1], t[2], t[-1])
````

````python
# Domanda: cosa viene stampato in questo caso?
print(t[2][-1]) # t[2] -> 'cool!'[-1]
````

Poiché le tuple sono dei tipi di dati che appartengono alla classe **SEQUENZE**, possiamo applicare gli stessi operatori che abbiamo visto per le stringhe:

````python
# Uso della funzione `len`
print(len(t))
````

````python
# Accesso diretto ad un elemento (operatore di indexing)
print(t[2])
````

````python
# Gli operatori di slicing
print(t[0:2])
````

````python
a = tuple([3])
print(type(a), a)
````

````python
help(tuple)
````

````python
a = ('cool',)
type(a)
````

````python
# Test di appartenenza
print('c' in t[2])
print(2.3 in t)
````

````python
# Supporto del costrutto <for>
for e in t:
    print(e)
````

````python
# Concatenazione
t2 = ('prova', 'prova')
print('concatenazione:', t+t2)
````

````python
# Repetition
print('repetition:', t2*3)
````

**NOTA:** In alcuni casi particolari, può essere utile definire una tupla di lunghezza pari a uno, ovvero contenente un singolo elemento. In questo caso, si deve usare la (strana) sintassi `(1,)`: si noti la virgola dopo l'uno.

````python
a=(1)
b=(1,)
print(type(a), type(b))
````

## 4.4 Le liste built-in 
Le **liste** di Python sono delle sequenze di elementi come le tuple, ma a differenza di quest'ultime, ogni singolo elemento di una lista può essere modificato senza dover creare una nuova lista. La sintassi per gestire le liste è simile a quella delle tuple: la differenza principale consiste nell'usare le parentesi quadre invece delle parentesi tonde.

Una possibile rappresentazione "box-and-pointer" di una lista è la seguente:

<img src="images/listaBoxPointer.PNG" alt="Stringa BoxAndPointer" style="width: 400px;"/>

I metodi **COSTRUTTORI** per le liste sono:

```
L1 = []
L2 = list()
L3 = [1, 2.3, 'cool!', False]
```

I metodi **SELETTORI** di una lista sono l'operatore di indicizzazione che restituisce l'$i$-esimo elemento della lista, e gli operatori di *slicing*, come visti per le stringhe e le tuple. Si vedano gli esempi sotto.

````python
Ls = [1, 2.3, 'cool!', False]
print(type(Ls), Ls)
````

````python
Ls[1:]
````

````python
print(Ls)
````

````python
# Lista vuota
print(len([]), type([]))
````

Anche per le liste possiamo usare tutte le funzioni che vengono solitamente definite in Python per le sequenze di elementi:

````python
# Uso della funzione `len`
print(len(Ls))
````

````python
# Accesso diretto ad un elemento
print(Ls[2])
````

````python
# Gli operatori di slicing
print(Ls[1:2])
````

````python
Ls = ['a','b','c','d']
print(Ls[1:3])
````

**NOTA:** A differenza delle tuple e delle stringhe, le liste si possono modificare: il metodo `L.append(e)` aggiunge l'elemento `e` in fondo alla lista `L`, e con l'operatore di indicizzazione si può assegnare un nuovo valore a un elemento della lista, come in `L[1] = 'z'`.

````python
Ls.append('e')
````

````python
Ls
````

````python
Ls[1] = 'z' 
print(Ls)
````

````python
# Test di appartenenza
print('c' in Ls)
print('2.3' in Ls)
````

````python
# Supporto del costrutto <for>
for e in Ls:
    print(e)
````

````python
# Concatenazione
L2 = ['prova', 'prova']
print('concatenazione:', Ls+L2)
````

````python
# Repetition
print('repetition:',L2*3)
````

Basta modificare uno dei due, che il cambiamento si riflette sull'altro: si parla in questo caso di **SIDE EFFECTS**, in quanto si potrebbero introdurre degli effetti non desiderati, molto difficili da individuare. Notare che problemi di questo tipo non si hanno con strutture dati di tipo **read only**.

### 4.4.2 List comprehensions
Quando abbiamo definito i plot di alcune semplici funzioni, abbiamo usato direttamente una sintassi di Python che permette di creare in modo molto semplice delle liste: le **list comprehensions**. Ad esempio:

````python
for e in range(10):
    print(e)
````

````python
S = [x**2 for x in range(10)]     # Map
V = [2**i for i in range(13)]     # Map
M = [x for x in S if x % 2 == 0]  # Filter
````

````python
print(S)
print(V)
print(M)
````

La stessa sintassi si può usare per le stringhe:

````python
Ls = "Vorrei, ma non posto!"
Ls = Ls.split(' ')
print(Ls)
````

````python
As = [(len(s), s.upper(), s.lower()) for s in Ls]
print(As)
````

## 4.5 Dizionari
I dizionari sono una delle strutture dati più utili e utilizzate di Python. I dizionari sono delle strutture dati che corrispondono, da un punto di vista logico, a delle liste di coppie **(chiave, valore)**, o, in termini tecnici, **(key, value)**. 

I dizionari sono degli oggetti **mutabili**, come le liste. Tuttavia, non sono delle sequenze: i valori non si selezionano con un indice intero, ma direttamente usando le loro chiavi (dalla versione 3.7 di Python, un dizionario mantiene l'ordine in cui sono state inserite le chiavi).

I metodi **COSTRUTTORI** per i dizionari sono i seguenti:

```
D1 = {}
D2 = dict()
D3 = {'a': 0, 'b': 1, 'c': 2}
```

Una chiave **(key)** di un dizionario viene utilizzata per indicizzare un dato valore **(value)**. Per esempio:

`D3["hello"] = "ciao"`

Abbiamo la chiave `hello` utilizzata per indicizzare l'elemento `ciao` nel dizionario `D3`. Per semplicità, potete pensare ai dizionari come se fossero delle liste di elementi indicizzati da altri oggetti **immutabili**, come ad esempio delle stringhe, delle tuple o dei numeri interi.

**ESEMPIO:**

````python
# Creo un dizionario vuoto
Vocabolario = dict()
print(type(Vocabolario), Vocabolario)
````

````python
Vocabolario["keep"] = "stai"
Vocabolario["calm"] = "sereno"
print("Enrico,", Vocabolario["keep"], Vocabolario["calm"])
````

````python
print(Vocabolario)
````

````python
# È possibile enumerare i dizionari con un ciclo for
Vocabolario["hello"] = "ciao"
for key in Vocabolario:
    print("chiave:", key, "- valore:", Vocabolario[key])
````

### 4.5.1 Funzioni definite sui dizionari
Le seguenti funzioni sono molto utili per usare i dizionari:

* `len(d)`: restituisce il numero di elementi nel dizionario `d`
* `d.keys()`: restituisce una lista (vista) delle chiavi del dizionario `d`
* `d.values()`: restituisce una lista (vista) dei valori del dizionario `d`
* `key in d`: restituisce `True` se la chiave `key` è nel dizionario `d`
* `d.get(key, value)`: restituisce `d[key]` se `key` è in `d`, altrimenti restituisce il valore `value`
* `d[key] = value`: associa il valore `value` alla chiave `key` nel dizionario `d`
* `del d[key]`: rimuove la chiave `key` dal dizionario `d`
* `for key in d`: itera sulle chiavi del dizionario `d`

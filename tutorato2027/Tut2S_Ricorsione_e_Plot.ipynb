**Esercizio 1.**
Nel sistema di Peano l'operazione di somma è definita ricorsivamente come
\begin{aligned}
\text{Somma} \colon \mathbb N \times \mathbb N &\to \mathbb N, \\
(n,0) &\mapsto n, \\
(n, m + 1) &\mapsto \text{Somma}(n, m) + 1.
\end{aligned}

Scrivere una funzione `Somma(n, m)` che presi in input due numeri naturali restituisca la loro somma usando la definizone di Peano.

Si possono usare i comandi `+` e `-` solo per aggiungere e sottrarre $1$.

```python
def somma(n, m):
  if m == 0:
    return n
  return somma(n, m - 1) + 1

somma(0, 4)
```

**Esercizio 2.**
 Come per la somma, il prodotto in un sistema di Peano è definito ricorsivamente come
     
\begin{aligned}
\text{Prodotto} \colon \mathbb N \times \mathbb N &\to \mathbb N, \\
(n,0) &\mapsto 0, \\
(n, m + 1) &\mapsto \text{Somma}(\text{Prodotto}(n, m), n).
\end{aligned}

Scrivere una funzione `Prodotto(n, m)` che presi in input due numeri naturali restituisca il loro prodotto usando la definizone di Peano.

Ancora, si possono usare i comandi `+` e `-` solo per aggiungere e sottrarre $1$.

```python
def prodotto(n, m):
  if m == 0:
    return 0
  return somma(prodotto(n, m - 1), n)

prodotto(6, 8)
```

**Esercizio 3.** Dopo aver importato i pacchetti necessari, rappresentare il grafico della funzione dell'oscillatore smorzato $$f(x)=e^{-\frac{x}{5}}cos(2\pi x).$$

 **Esercizio 5.**
 La serie di Leibniz è una serie infinita che converge a $\frac{\pi}{4}$ definita
$$ \frac{\pi}{4}=\sum _{i=0}^{+\infty} \frac{(-1)^i}{2i+1}=1-\frac{1}{3}+\frac{1}{5}-\frac{1}{7}+...$$
Implementare una funzione `pi_leibniz(n)` che dato un numero naturale `n` restituisce l'approssimazione di $\pi$ dopo `n` somme della serie.

```python
def pi_leibniz(n):
  def rec(x,i):
    if i==n:
      return 4*x
    return rec(x+((-1)**i)/(2*i+1),i+1)
  return rec(0,0)

pi_leibniz(800)
```

**Esercizio 6.** Il coeficiente binomiale gode della proprietà ricorsiva $$\binom{n}{k}=\binom{n-1}{k-1}+\binom{n-1}{k}.$$
Con i casi base: $\binom{n}{0}=1$ e $\binom{n}{n}=1.$

Implementare la funzione ricorsiva `binomiale(n,k)` che calcoli il coeficiente binomiale $\binom{n}{k}$ usando esclusivamente la formula ricorsiva senza usare moltiplicazione o fattoriale.

```python
def binomiale(n, k):
  if k == 0:
    return 1
  if n == k:
    return 1
  if k > n or k < 0:
    return 0
  return binomiale(n - 1, k - 1) + binomiale(n - 1, k)

binomiale(5, 2)
```

**Esercizio 8.** Scrivere due funzioni ricorsive `pari(n)` e `dispari(n)` che ritornino un valore booleano in base a se `n` è rispettivamente pari o dispari senza usare divisioni, resti, quozienti e simili.

```python
def pari(n):
  if n == 0:
    return True
  return dispari(n - 1)

def dispari(n):
  if n == 0:
    return False
  return pari(n - 1)
```

**Esercizio 9.**  Un'onda quadra può essere approssimata dalla seguente serie
$$
    \frac4\pi\sum_{k=1}^{+\infty} \frac{\sin(\frac{k\pi}2)}{k}\cos(kx).
$$

Scrivere un programma che dato un intero $n$ stampi il grafico con l'approssimazione della serie dopo $n$ somme.

Usare le stesse funzioni per stampare un grafico con le funzioni sovrapposte date dalle approssimazioni per $2^i$ con $i=0,1, \dots, n$. (Conviene scegliere valori molto bassi di $n$).


```python
from numpy import linspace
from math import pi, sin, cos
from matplotlib.pyplot import plot, show, legend

n=100
m=7

def quadra(x, n):
  def rec(y ,k=1):
    if k == n:
      return 4/pi*y
    return rec(y + (sin(k*pi/2)*cos(k*x)/k), k + 1)
  return rec(0)

X = linspace(-2*pi, 2*pi, 100000)
Y = [quadra(x, n) for x in X]
plot(X,Y)
show()

for i in range(0,m):
  Y = [quadra(x, 2**i) for x in X]
  plot(X,Y, label=f'2^{i}')
  legend()
show()
```

**Esercizio 10.**  
Con rifermiento all'esercizio sulla congettura di Collatz dello scorso tutorato, scrivere un programma che prenda in input un numero naturale `n` mediante la funzione `int(input())` e


1.   calcoli dopo quanti passaggi la successione di `n` raggiunge $1$;
2.   calcoli il numero più alto raggiunto dalla successione;
3.   stampi il grafico con sulle ascisse l'intervallo naturale da $0$ al numero di passaggi totale e sulle ordinate il valore assunto dalla successione in tale punto.

*Consiglio: Usa una funzione `collatz(n, i)` che calcola l'`i`-esimo termine della successione di `n`.*


```python
from matplotlib.pyplot import plot, show

def collatz(n, x):
  if x == 0:
    return n
  if n%2 == 0:
    return collatz(n/2, x - 1)
  return collatz(3*n + 1, x - 1)

def len_collatz(n):
  def count(n, i=0):
    if n == 1:
     return i
    if n%2 == 0:
      return count(n/2, i + 1)
    return count(3*n + 1, i + 1)
  return count(n)

n = int(input('Inserisci un numero: '))
l = len_collatz(n)
X = [x for x in range (0, l+1)]
Y = [collatz(n, x) for x in X]
m = int(max(Y))
print(f'La serie di {n} raggiunge 1 dopo {l} passi, raggiungendo un picco di {m}')
plot(X,Y)
show()
```

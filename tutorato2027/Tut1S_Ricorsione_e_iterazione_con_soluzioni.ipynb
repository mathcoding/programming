Essa postula che per ogni $n \in \mathbb{N}$ la successione $\{a_m\}$ costruita come

$$a_0 = n, \quad a_{i+1} = \begin{cases} a_i/2 & a_i \text{ pari}, \\ 3a_i + 1 & a_i \text{ dispari}. \end{cases}$$

raggiunga ad un certo punto il valore 1.

**Esercizio 3.** L'algoritmo di Euclide è un metodo per il calcolo del massimo comune divisore. Dati $a$ e $b$ , se $b = 0$ allora $a$ è il massimo comune divisore tra i due. Altrimenti, si calcola il resto $r$ della divisione di $a$ per $b$. Se $r = 0$ l'algoritmo termina e $MCD(a,b) = b$, altrimenti occorre porre $a' = b$ e $b' = r$ ripetendo nuovamente l'algoritmo per $a'$ e $b'$ e così via finchè non termina.

Implementare una funzione ricorsiva `mcd(a, b)` che presi in input due numeri positivi a e b, restituisca il loro massimo comune divisore.

*Domanda: Cosa ci assicura che l'algoritmo termini?*



```python
def mcd(a, b):
  if b == 0:
    return a
  return mcd(b, a%b)

mcd(4, 8)
```

**Esercizio 4.** Scrivere una funzione `rad(x, tol=1e-07)` che preso in input un numero positivo x ne calcoli la radice quadrata per bisezione con una tolleranza pari a tol.

`tol=*` assegna un valore `*` di default alla variabile `tol`, questo vuol dire che passando a `rad` solo il parametro `x` viene calcolato `rad(x, *)`.

**HC** Si provi ora a riscrivere la funzione senza utilizzare gli operatori `*` e `**` per prodotto e elevamento a potenza.

```python
def rad(x, tol=1e-07):
  def rec(a, b):
        y = (a + b)/2
        if abs(b - a)/2 < tol:
            return y
        y2 = y*y
        if abs(y2 - x) < tol:
            return y
        if y2 < x:
            return rec(y, b)
        return rec(a, y)
  return rec(0, x)

rad(17)
```

**Esercizio 5.** Il metodo di Newton per trovare la radice quadrata di un numero si basa sul fatto che se $y_k \ge 0$ è un'approssimazione della radice quadrata di $x \ge 0$, allora un'approssimazione migliore è data da
$$y_{k+1} = \frac{y_k + \frac{x}{y_k}}{2}$$
Implementare una funzione `n_rad(x, tol=1e-07)` che preso in input un numero positivo x ne calcola la radice quadrata con tolleranza pari a tol mediante il metodo di Newton.

```python
def n_rad(x, tol=1e-07):
  def rec(y):
        z = 0.5*(y + x/y)
        if abs(z - y) < tol:
            return z
        return rec(z)
  if x == 0:
        return 0
  return rec(x)

n_rad(17)
```

**Esercizio 8.** Implementare una funzione ricorsiva `bisect(f, a, b, tol=1e-04)` che calcoli gli zeri della funzione `f` nell'intervallo di estremi `a` e `b`.

```python
def f(x): return x**3 + x - 2

def bisect(f, a, b, tol=1e-04):
    if abs(f(a)) < tol:
        return a
    if abs(f(b)) < tol:
        return b
    if abs(a - b) < tol and abs(f(a)) > tol:
        return None
    m = (a + b)/2
    if abs(f(m)) < tol:
        return m
    if f(a)*f(m) < 0:
        return bisect(f, a, m)
    return bisect(f, m, b)

bisect(f,10,-10)
```

**Esercizio 10.** La funzione di Ackermann è un esempio più noto di funzione ricorsiva totale non primitiva ed è definita come

$$f_x(y, z) = \begin{cases}  z & x = 0, y = 0 \\ f_0(y - 1, z) + 1 & x = 0, y \neq 0 \\ 0 & x = 1, y = 0 \\ 1 & x \ge 2, y = 0 \\ f_{x-1}(f_x(y - 1, z), z) & x \ne 0, y \ne 0  \end{cases}$$

Scrivere una funzione ricorsiva `f(x, y, z)` tale che
$$f(x, y, z) = f_x(y, z).$$

```python
def f(x, y, z):
  if x == 0:
    if y == 0:
      return z
    return f(x, y - 1, z) + 1
  if y == 0:
    if x == 1:
      return 0
    return 1
  return f(x - 1, f(x, y - 1, z), z)

f(1,2,6)
```
